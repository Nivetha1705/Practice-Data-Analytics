```python
#For Loop
for i in range(1,6):
    print(i*2, end=" ")
```

```python
for i in range(5):
    print("*", end=" ")
```

```python
n=int(input("Enter the number:"))
for i in range(2,11,2):
    if n==i:
        print("Number found")
```

```python
n=int(input("Enter the number:"))
for i in range(2,11,2):
    if n%i==0:
        print("Number found")
        break
    else:
        print("Number not found")
        break
```

```python
#it will only print 0 and then it comes out of the loop
for i in range(5):
    print(i)
    break
```

```python
for i in range(5):
    if i==3:
        continue
    print(i)   
```

```python
fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit)
```

```python
student = {"name": "Priya", "age": 25}
for key, value in student.items():
    print(key, ":", value)
```

```python
#While loop

i=1
while i<=5:
    print("*", end=" ")
    i+=1
```

```python
#Break statement
for i in range(10):
    if i == 5:
        break
    print(i, end=" ")
```

```python
#Continue statement
for i in range(10):
    if i % 2 == 0:
        continue
    print(i, end=" ")
```
