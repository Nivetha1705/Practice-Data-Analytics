Arrays in numpy
ARRAYS-continous memory allocation
Numpy always uses static array and stores only homogenous datas

We have to store the data in the array to use in numpy. So we have to understand that there are 3 types of arrays in numpy
1D array - Vector (only columns)
2D array - Matrix or matrice (Rows and Columns)
3D array - Tensor (Blocks, Rows and Columns)

Before creating an array, we have to observe 2 things
1.Dimensionality
2.Homogenous or Heterogenous datas

#1.Create an array to store the ages of 5 students --> This gives only one piece of information, do 1D array
---> 1 Dimensional Array
---> Homogenous datas

#2.Create an array to store the ages of students belonging to 2 classrooms with 5 students each. --> Gives 2 piece of information,so its 2D array
---> 2 Dimensional Array
---> Homogenous datas
Real world datasets are in the form of matrix only

#3.Create an array to store the ages of students belonging to2 schools having 3 classrooms with 5 students each. --->Gives 3 piece of information, so its 3D array
---> 3 Dimensional Array
---> Homogenous datas

3D array is a combination of multiple 2D arrays combined together

```python
#1.using List --1D array
import numpy as np
l=[1,2,3,4,5]
a=np.array(l)
print(a)

#or

a=np.array([1,45,67,89,39])
print(a)
```

```python
import numpy as np
a=np.array([[78,89,28,40,28],[27,79,90,28,67]])
print(a)
```

```python
#2.using arange() --1D array
a=np.arange(2,22,2)
print(a)
```

```python
#3.using zeros()  --> 2D array. This takes parameters only as tuple
a=np.zeros((2,2))
print(a)   #Stores data in the of float
```

```python
#4.using ones()
a=np.ones((3,3))
print(a)
```

```python
#5. using linspace()
import numpy as np
arr=np.linspace(0,1,5)
print(arr)
```

```python
#6. using rand()
import numpy as np
arr=np.random.rand(5)
print(arr)
```

When we create an array in numpy, an array object is being created. This array object consists of the array that has been created along with some instance variables and also methods together called as attributes

The types of instance variables present are
1.shape
2.size
3.ndim
4.dtype

```python
#instance variables inside the array object
#Shape
l=[1,2,3,4,64,89]
a=np.array(l)
print(a.shape)

```

```python
#Size
l=[1,2,3,4,64,89]
a=np.array(l)
print(a.size)
```

```python
#dtype
l=[1,2,3,4,64,89]
a=np.array(l)
print(a.dtype)
```

```python
#ndim
l=[1,2,3,4,64,89]
a=np.array(l)
print(a.ndim)
```

```python
import numpy as np
a=[[1,2,3,],[1,2,3]]
b=np.array(a)
print(b)
print(b.shape)
print(b.size)
print(b.ndim)
print(b.dtype)
```

```python
a=np.arange(2,22,2)
print(a)
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

```python
a=np.zeros((2,2))
print(a) 
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

```python
a=np.ones((3,3))
print(a)
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

Sometimes we need only some data from the datasets so indexing and slicing would help to get this

```python
arr=np.array([1,3,5,7,9,2,4,6,8])
print(arr[7])
```

```python
import numpy as np
arr=np.random.rand(100)
print(arr[0:101:1])
print(arr[34:78:4])
print(arr[99:52:-2])
print(arr[::-2])
```
