FileHandling - READ
Every time when the file is opened to read or write then close it.

```python
#Open and read the file
f=open("Filehandling.txt","r")   # r denotes mode
print(f.read())
```

```python
#Another way to open the file so that python automatically closes the file if we forget to do it
with open("Filehandling.txt","r") as f:   #parameters(filename,mode)
    print(f.read())
```

```python
#Once the file is read,close it
with open("Filehandling.txt","r") as f:
    print(f.read())
    f.close()
```

```python
#To read the number of characters as we need
f=open("Filehandling.txt","r") 
print(f.read(5))

f=open("Filehandling.txt","r") 
print(f.read(10))
```

```python
#To read the lines
with open("Filehandling.txt","r") as f:
    print(f.readline())  # 1 line
    
with open("Filehandling.txt","r") as f:
    print(f.readline())    #2 lines
    print(f.readline())
```

```python
#Looping in file
f=open("Filehandling.txt","r")
for i in f:
    print(i)
```

```python
#We can also specify the file to handle whether text or binary
with open("Filehandling.txt","rt") as f:   #rt-read text
    print(f.read())
    f.close()
```

Filehandling - Write
To write to an existing file, you must add a parameter to the open() function:

"a" - Append - will append to the end of the file

"w" - Write - will overwrite any existing content

```python
#This will overwrite all the contents present in the file
f=open("Filehandling.txt","w")
f.write("qwertyuiopasdfghjklzxcvbnm")

with open("Filehandling.txt","r") as f:
    print(f.read())
```

```python
#Append will add the content at the end of the file
f=open("Filehandling.txt","a")
f.write("Data analysis is the process of collecting, cleaning, organizing, examining, and interpreting data to discover useful information, 
identify patterns, and support decision-making.")

with open("Filehandling.txt","r") as f:
    print(f.read())
```
