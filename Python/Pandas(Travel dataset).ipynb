MISSING VALUES
We cannot ignore missing datas and analyze the dataset as it is. If we do so, the conclusion will not be correct. So handling the missing values is very important.

```python
#First we have to check whether there is any missing values present in the dataset, by using isnull() which is a method.
df.isnull()  #Gives True or False (If True-->missing value, False-->Data is present)
```

```python
df.isnull().sum()  #sum() Gives the count of missing values for each columns completely.
```

Different ways of handling missing values in pandas
1. dropna()
2. fillna()

DROPNA() - It is a method present inside the dataframe 
If the rows contain any null values the entire row will be dropped in the dataframe.
drops all the rows containg null values in it.
When dropna() is used, it creates a copy of the original dataframe and the rows will be dropped in the copy, keeping the original dataframe safe.

```python
df.dropna()
```

```python
df=df.dropna()
df.isnull().sum()
```

```python
#Initially there was 4888 rows in the table, and now we can check how many rows are there after dropping roews with null values
print(df.shape)  #Only 4128 are present after dropping rows with null values.
#But if so many rows are removed like this, the conclusion we get after performing Data analysis would not be correct.
```

Till this, dropna() created a new data frame and made changes in that. If we need the changes to be done in the same dataframe (or) excel sheet,
INPLACE should be given as parameter and the default value False should be changed to True

```python
df.dropna(inplace=True)   #This makes changes in the original dataframe itself.
df.isnull().sum()
```

```python
df.shape
```

FILLNA()
Whereever there are missing values, we need to fill it with some datas, instead of dropping the rows entirely.
Then we should use Fillna() method. In fillna() also inplace works like same as dropna().
string columns cannot accept integer values 0 tobe filled. So we have to identify only interger columns and fill 0 there

```python
df.fillna(0,inplace=True)
print(df)
```

```python
#FILLNA()

num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_columns].fillna(0)
print(df)
```

```python
cat_cols = df.select_dtypes(exclude='number').columns
df[cat_cols] = df[cat_cols].fillna('Unknown')
print(df)
```

Concatenation of two dataframes - Vertically attached
In the below example, we have to get the total sales of both july and aug month which are in two separate dataframes.
This concat() attaches two datasets in a new dataframe.
Here we have combined 2 dataframes vertically, one below the other

```python
import pandas
Julydata={
    'Products':["iphone","ipad","Mac","Tablet"],  
    'Total':[100,99,100,98]
}
Julysales=pd.DataFrame(Julydata)

import pandas
Augdata={
    'Products':["iphone","ipad","Mac","Tablet"],  
    'Total':[250,199,46,68]
}
Augsales=pd.DataFrame(Augdata)

result=pd.concat([Julysales,Augsales])
result
```

Merging of datasets - Horizontally Attached

```python
import pandas as pd
data1={
    "Id" : [1,2,3,4],
    "Name" : ["Raj", "Rishi", "Khamal", "Eniyan"]
}
df1=pd.DataFrame(data1)

data2={
    "Id" : [1,2,3,4],
    "Grade" : ["A", "C", "B", "A"]
}
df2=pd.DataFrame(data2)

result=pd.merge(df1,df2, on = 'Id')
result
```

Aggregate functions on Pandas - Aggregation should be performed only on numerical columns.
1. count()
2. sum()
3. mean()

```python
data={
    "Subject" : ["Maths","Science","Social", "English","Tamil"],
    "Marks" :[100,100,99,56,78]
}
raju=pd.DataFrame(data)
```

```python
#Count()
count=raju["Marks"].count()
count

#or

m=raju["Marks"]
c=m.count()
print(c)

#or
raju["Marks"].count()
```

```python
#Sum()
sum=raju["Marks"].sum()
sum

#or

m=raju["Marks"]
s=m.sum()
print(s)

raju["Marks"].sum()
```

```python
#Mean()
mean=raju["Marks"].mean()
mean

#or

m=raju["Marks"]
mean=m.mean()
print(mean)

raju["Marks"].mean()
```
