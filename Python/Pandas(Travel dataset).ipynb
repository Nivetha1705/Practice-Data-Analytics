In the above cells we could see all the rows till the end of the dataframe is getting printed. so we can specify the rows that we need also by using the method called as loc. By using loc both indexing and slicing can be done

LOC and ILOC

```python
#loc is an instance vaiable which denotes the location
import pandas as pd
df=pd.read_csv("Travel.csv")
print(df.loc[0])   # Similar to indexing in python
```

```python
#This is similar to slicing in python by using loc
import pandas as pd
df=pd.read_csv("Travel.csv")
print(df.loc[0:2])  #Prints the first 3rows 
```

```python
df.loc[0]
```

```python
df.iloc[0]
```

In the both above cells, we could see loc[0] and iloc[0] gives the same output of only the 0th index that is 1st row.

```python
df.loc[0:2]  #here in slicing in loc 0,1,2 (3 rows) will be printed where 2 is also included
```

```python
df.iloc[0:2]  #here in slicing in iloc only 0,1 (2 rows) will be printed where 2 is excluded here
```

CONDITIONAL FILTERING OR BOOLEAN FILTERING

```python
#Age of people which is greater than 40 
df[df["Age"]>40]   #To get the entire dataframe we should give it inside [] brackets
```

```python
df["Age"]>40   #If we do not give it inside the [] brackets, we will only get whether it is true or false and not the entire dataframe
```

```python
#Filtering using 2 conditions - Age and CityTier
df[(df["Age"]<35) & (df["CityTier"]==1)]
```

```python
#Filtering using age,gender and occupation at same time
df[(df["Age"]>=25) & (df["Age"]<=40) & (df["Gender"]=="Male") & (df["Occupation"]=="Salaried")]  
```

SORTING OF DATAS
The raw datas will not be arranged in any order properly. So we need to sort it either in asc or desc order
asc --> Lower - Higher
desc--> Higher - Lower

Sort_values(by) is a method in the dataframe that holds a default parameter called as 'by' where we should give the column name by which it should be sorted.

```python
df.sort_values(by="Age")
#or
df.sort_values("Age")   #This gives the output of age that is sorted in ascending order by default.
```

```python
#def sort_values(by, ascending=True) --> default parameter=True 
df.sort_values("Age", ascending=False)  _#Here we gave false, and hence the age will be sorted in desc order.
```

MISSING VALUES
We cannot ignore missing datas and analyze the dataset as it is. If we do so, the conclusion will not be correct. So handling the missing values is very important.

```python
#First we have to check whether there is any missing values present in the dataset, by using isnull() which is a method.
df.isnull()  #Gives True or False (If True-->missing value, False-->Data is present)
```

```python
df.isnull().sum()  #sum() Gives the count of missing values for each columns completely.
```
