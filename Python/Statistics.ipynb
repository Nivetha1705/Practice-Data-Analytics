TWO more terminologies to understand :
1.Random Sampling - A probability method where a researcher picks a subset of individuals from a larger group so that every individual gets an equal and exact chance of being chosen.
2.Bias - Biased towards one type of data but we cannot say it's correct or wrong.

SAMPLE has two types of sample:
1. Probability sample  --> 4 types - simple random sampling, systematic sampling, stratified sampling, clusterd sampling
2. Non Probability Sample 

```python
#1.Simple random sampling:
for seed in [1,2,3,4,5]:
    rand=df.sample(30,random_state=seed)
    print(seed, rand['salary'].mean())
print("Mean: ", df['salary'].mean())   
#when the same seed 1 is given, the same 30 random employees will be got.
```

```python
#Instead of picking 5 random seeds with 30 employees each, we can show favourism here(Bias)
small = df.nsmallest(5, "salary")
print("mean :", small["salary"].mean())
```

Three rules to use:
If the numbers are evenly spread - mean
If numbers are few extreme - median
For words/categories based columns - mode

```python
#when the values are not equally distributed we cannot report the mean or median on its own, we have to find the spread.
#Spread - How far the values are from the average.
#To find the spread, range is used which takes two extreme vlues (max-min)
print(df["salary"].describe().round(2))
sal=df["salary"]
print(sal.max() - sal.min())
#This huge diff is because of that one person with 52 lakhs salary.
```

```python
#To overcome the above problem, we can use quantiles -->25%, 50%, 75%

q1=(sal.quantile(0.25))
print(q1)   #This means that 25% of the people are earning below 59003
q2=(sal.quantile(0.50)) 
print(q2)#This means that 50% of the people are earning below 68647 (Median that is 50%)
q3=(sal.quantile(0.75))
print(q3)#This means that 75% of the people are earning below 80599
```

```python
#Interquantile range = q3 - q1
iqr = q3 - q1
print(iqr)  

#This is the diff between range and IQR
#Range = 5166915, IQR = 21596 (Practically possible)
```

Paring :
When it's evenly distributed, use Mean + Range (Both uses evey values and gets wreked by extremes)
When it extremely distributed, use Median + IQR (cares about positions and ignore's extremes)

```python
#MEAN - MEDIAN = SKEW
print(sal.mean() - sal.median())
# 3 types of skew - positive, negative, neutral
#lopsided means curve
```

```python
print(df["salary"].mean()-df["salary"].median())
print(sal.skew())  #Positive skew, this is because that one employee with 52 lakhs salary.
```

#Every problem will be sorted if we remove that one person called as OUTLIER. Should we remove them ?
If we dlt the outlier - problem gets fixed
or It may be a salary of CEO also where we cannot remove it.

WHY a value can be extreme ?
1.A mistake - fix it or remove
2.Real and it matters - the CEO earns 52 lakhs. Keep it and report the median, not mean
3.Real and it is to the point

You may remove a value - But not quiently. You should always ask, WHY?, What? and HOW MANY?

```python
#Whenever we need to remove the outlier, we have to know the range. Anything outside the fence is called as outliers.
#Anything outside the fence gets flagged and can be removed immediately. --> BOX PLOT
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
print(lower_fence)
print(upper_fence)
```

```python
print(df[(sal<lower_fence) | (sal>upper_fence)])  #2 outliers gets flagged
```

```python
#Similarly on age column
age=df["age"]
q1,q3=age.quantile([0.25,0.75])
print(q1,q3)

iqr=q3-q1
print(iqr)

low=q1-1.5*iqr
upp=q3+1.5*iqr
print(low,upp)

print(df[(age<low) | (age>upp)])

#Age cannot be -4 and 210. These are fraud datas, and it should be removed.
```

The same rule has been followed for the age column. It matters on which column we are following the box plot.
Salary can be real with 52 lakhs, but the age is not possible.

 AND for AGE - FIX

#How would you measure spread using every single values ?  --> MEan Absolute Deviation
Step 1 - For each value find out how far it's from the mean.
Step 2 - Remove the negative signs and make it absolute value, and find the average (spread)
This method is called as Mean absolute deviation, but almost nobody uses it.

One more way of doing this is,
In step 2- Square the values and fine the average. then it's called as Variance

We squared the value and got the variance. But what's the unit rupees square, doesnt makes any sense.
so we take square root of variance which is called as STANDARD DEVIATION. NOW rupees square - rupees

Standard Deviation depicts how much the values are deviated from the mean value, making use of every single value.

```python
import numpy as np

print("Numpy: ", np.std(sal))  # Takes n(population)
print("Pandas: ", sal.std())  #Takes n-1 (Sample data)
```

```python
print("Numpy: ", np.std(sal, ddof =1)) #Making numpy to take sample data. ddof=delta degree of freedom
#It's always better to use sample data, so if we set ddof=1 then it takes sample or defaulty it takes population in numpy
```
