POPULATION

Getting all the actual datas
Eg : ALL students of TAP Academy
What we fetch from Population is called as PARAMETER

SAMPLE DATA - Taking a small sample and assuming it would be equivalent to the answer

Subset of a population is sample data
Eg : Only certain batches of TAP Academy
What we fetch from sample data is called as STATISTIC

TWO more terminologies to understand :
1.Random Sampling - A probability method where a researcher picks a subset of individuals from a larger group so that every individual gets an equal and exact chance of being chosen.
2.Bias - Biased towards one type of data but we cannot say it's correct or wrong.

SAMPLE has two types of sample:
1. Probability sample  --> 4 types - simple random sampling, systematic sampling, stratified sampling, clusterd sampling
2. Non Probability Sample 

```python
#1.Simple random sampling:
for seed in [1,2,3,4,5]:
    rand=df.sample(30,random_state=seed)
    print(seed, rand['salary'].mean())
print("Mean: ", df['salary'].mean())   
#when the same seed 1 is given, the same 30 random employees will be got.
```

```python
#Instead of picking 5 random seeds with 30 employees each, we can show favourism here(Bias)
small = df.nsmallest(5, "salary")
print("mean :", small["salary"].mean())
```

#2.Systematic sampling - Order is followed here.
out of 2000 products 200 products have to be picked and mean of those products will be found. 
2000/200 = 10. So 1,11,21,31,41 these ways products will be picked till 2000.
10 machines are there. If 1 machine produces 1 products and all 1,11,21,31,41 belongs to machine 1, so systematic sampling fails here.

```python
#Stratified sampling
Total staffs-2000. Tech - 1200, Sales-500, HR- 300. We have to pick 200 out of them.
It might randomply pick the same department employees again and again. To avoid that,
Tech - 1200/2000 = 0.6 = 60%,similarly Sales = 25% and HR=15%

60% of 200 = 120, 25% of 200 =50, 15% of 200 = 30
Total we got 200 equally distributed for each department, so that we will get the answer accurate.
```

#Cluster Sampling:
