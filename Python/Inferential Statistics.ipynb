```python
#Avg salary of both Tech and HR department

t=clean[clean.department=="Tech"].salary
hr=clean[clean.department=="HR"].salary
print("TECH:",len(t), round(t.mean()))
print("HR:",len(hr), round(hr.mean()))

print("Difference: ", round(t.mean() - hr.mean(),2))  #25000, Is this real or just because of the sample we took?
```

The above difference makes a meaning by doing a testing called as HYPOTHESIS TESTING
It tells whether to rely on or ignore the data.
    Step 1 : Assume there is no diff btw them - NULL HYPOTHESIS H0
    Step 2: Assume there is a diff btw them -ALTERNATE HYPOTHESIS H1
To represent this in maths, we have p-value. 
Optimal p-value is 0.05(threshold) -->Between 0 and 1
p<0.05 - reject H0
p>0.05 - Failed to reject H0(No enough evidence to prove)

```python
sample=clean.salary.sample(100, replace=True, random_state=54)
from scipy import stats
t,p = stats.ttest_1samp(sample,65000)
print("Sample Average: ", sample.mean())
print("The claim: ", 65000)
print("P-value: ", round(p,2))  #reject H0 beacuse p<0.05
```

```python
tech=clean[clean.department=="Tech"].salary
hr=clean[clean.department=="HR"].salary

t,p = stats.ttest_ind(t,hr, equal_var =False)
print("TECH:", round(tech.mean()))
print("HR:", round(hr.mean()))
print("P-value: ", round(p,2))   #very very less value, so reject H0
```

```python
from scipy import stats
marketing = clean[clean.department == "Marketing"].salary
sales = clean[clean.department =="Sales"].salary

t,p = stats.ttest_ind(marketing,sales, equal_var = False)
print("Marketing Mean:", round(marketing.mean(),3))
print("Sales Mean:", round(sales.mean(),3))
print("P-value: ", round(p,3))   #Less than 0.05, so reject H0
```

```python
#A/B Testing - Same as Chi-square but on [2*2] matrix
#version A-96
#version B-138      #1104,96
                    #1062,138  out of 1200
```

```python
import numpy as np
from scipy import stats
ab=np.array([[1104,96],
            [1062,138]])
print("A:", round(100*96/1200,1),"%")
print("B:",round(100*138/1200,1),"%")
chi2,p,dof,exp = stats.chi2_contingency(ab, correction=False)
print("P-value:", round(p,4))   #<0.05 (reject H0)
```

```python
small=np.array([[110,10],     #120 in total
               [106,14]])
print("A:",round(100*10/120,1),"%")
print("B:",round(100*14/120,1),"%")
chi2,p,dof,exp =  stats.chi2_contingency(small,correction=False)
print("P-value:",round(p,4))    #>0.05 (Failed to reject H0)

#here when the sample reduced to 120 people, the p value got more. Hence we should know the no.of people required for sample.
```
