All the employees who are either in tech or have rating as excellent


```python
t = df.department=="Tech"
e = df.rating =="Excellent"
print(t.mean())
print(e.mean())
print("Added Together: ", (t.mean() + e.mean()).round(2))
```

```python
#One more way of achieving this answer by using OR operator
print("OR Operator: ", (t|e).mean().round(2))

#This is the accurate result because itremoves the intersection between the two columns[People who are in tech having Excellent Rating]. it should 
#occur only one  P(t or e)=p(t) + p(e) - p(t intersection e)
```

All employees who are working in Tech and having Excellent Rating


```python
print("Mul together: ", (t.mean() * e.mean()).round(2))
print("And Operator: ", (t&e).mean().round(2))
#This would be almost same for both
```

CONDITIONAL PROBABILITY

```python
#P(A|B) -- > A given B which means first B will be evaluated and based on the data we get, A will be evaluated
r = (df.rating=="Excellent").mean().round(3)
print(r)
t= (df[df.department=="Tech"])     
print("P(Excellent|Tech): ", (t.rating == "Excellent").mean().round(3))

#This first finds out all the people who belongs to Tech and then applies Excellent rating over them. 
```

```python
#Now reverse, P(B|A)
t = (df.department=="Tech").mean().round(3)
print(t)
r=(df[df.rating=="Excellent"])
print("P(Tech|Excellent): ", (r.department == "Tech").mean().round(3))

#Both answers came the same because of coincidence
#Both cannot be same values because when we take department first there can be 49 people and when we take rating first there can only be 40 people, and then when we filter the answer will be different only and not the same.
```
