OPERATORS IN PYTHON 
-Arithmetic operators --> +,-,*,/,//,%,**
-Logical operators --> and, or, not
-Assignment Operators --> =,+=,-=,*=,/=,//=,%=,**=
-Comparison or Relational Operators --> ==,!=,>,<,>=,<=
-Bitwise Operators --> &,|,~,^,<<,>>
-Membership Operators --> in, not in
-Identity Operators --> is, is not

```python
a=10
b=20
c=a+b
print(c)
```

```python
a,b=2,3
a**b
```

```python
a=3
a**=2 #a=a**2 --> 3**2 = 9 
print(a)
```

```python
a,b=7,5
not(a<b)
```

```python
#ARITHMETIC OPERATORS

a,b=2,1
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
#LOGICAL OPERATORS

a,b=12,45

print(a>b and b>a)

print(a<b or a>b)

print(not(a<b))
```

```python
#ASSIGNMENT OPERATORS

a=67
a+=24
print(a)
a-=24
print(a)
a*=24
print(a)
a/=24
print(a)
a//=24
print(a)
a%=24
print(a)
a**=24
print(a)
```

```python
#COMPARISON OR RALATIONAL OPERATORS

a=56
b=90
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
print(a>b)
print(a<b)
```

```python
#BITWISE OPERATORS

a,b=45,34
print(a&b)
print(a|b)
print(a^b)
print(a<<b)
print(a>>b)
```

```python
#MEMBERSHIP OPERATORS

print(2 in [1,2,3])
print('Ta' in 'TAP')

print('L' not in 'Angel')
```

```python
#IDENTITY OPERATORS

a=23
b=23
print(a is b)

a=34
b=45
print(a is b)
```
