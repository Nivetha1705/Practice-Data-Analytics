Disadvantages of numpy:
Numpy don't have any labels. --> arr=np.array([22,21,34,56])
It doesn't have any label like it is a list of ages in numpy.
Without any information it is difficult for us to analyze what data it is.
But if it is in the form of tables with rows and columns, it will be easy for us to analyze the data.

PANDAS  (Alias name of pandas is pd)
WES MCKINNEY felt difficult while analyzing the financial datas. So he built pandas on top of numpy. 
The raw datas are arranged in tabular form and labels are given to it.Pandas was released in 2009.

Python-->Gives only raw data
Numpy -->Datas only in the form of numbers
Pandas-->Data is present in the table and every data should have label so that we can analyze it in a better way.

```python
import pandas
data={
    'Name':["Tim","Raj","Rishi","Nithish"],  #Multiple datas given inside list and every fature of list can be accessed here like duplicate values,heterogenous datas,etc
    'Age':[25,49,21,25]
}
df=pandas.DataFrame(data) #In pandas we should not tell as table, it is called "DataFrame".
print(df)   #In the output, the indexes would start from 0 as default.
```

```python
import pandas as pd
data={
    "id" : [1,2,3,4,5,6,7,8],
    "Name" : ['Kshitij','Sharath','Somanna','Poovizhi','Kshitij','Sharath','Somanna','Poovizhi'],
    "Salary" : [1000000,20000,34384,23950,1000000,20000,34384,23950],
    "Dept" : ['IT',"SALES","FIN","HR",'IT',"SALES","FIN","HR"]
    }
df=pd.DataFrame(data)
print(df)

#or

df=pd.DataFrame({
    "id" : [1,2,3,4,5,6,7,8],
    "Name" : ['Kshitij','Sharath','Somanna','Poovizhi','Kshitij','Sharath','Somanna','Poovizhi'],
    "Salary" : [1000000,20000,34384,23950,1000000,20000,34384,23950],
    "Dept" : ['IT',"SALES","FIN","HR",'IT',"SALES","FIN","HR"]
    })
print(df)

```

PICKING OF DATA - Before analyzing the data it is very importatnt
When we perform the below 5 things in pandas then it is called as EXPLORATORY DATA ANALYSIS (EDA)
1.info()
2.head()
3.tail()
4.shape
5.describe()

```python
#There is a method in pandas called as info() - which gives the entire information about the dataframe that is present inside the object
df.info()
```

```python
#head() - This is method in pandas to get the first 5 rows of the dataframe to analyze instead of getting crores of rows present in the table which would be easy to analyze.

df.head()
```

```python
#tail() - This is method in pandas to get the last 5 rows of the dataframe to analyze instead of getting crores of rows present in the table which would be easy to analyze.

df.tail()
```

```python
#Shape is an instance variable in pandas which tells the no.of rows and columns present in the dataframe. (rows,cols)
df.shape
```

```python
#describe() -- Describe is a method that is used for statistical analysis of numerical columns present in the dataframe.
df.describe()
```

SERIES --> It is also a class in pandas.
Series is one strip of data which is called as Vector in Numpy.
DataFrame is with rows and columns in pandas which is called as matrice in Numpy.

```python
#Series -->falls on 1D array (1 piece of data)
import pandas as pd
s=pd.Series(data=[22,23,24,25,78])
print(s) #Index starts with 0 by default
```

```python
#We can also specify our own index
import pandas as pd
s=pd.Series(data=[22,23,24,25,78],index=["Tim","Alex","Raju","Diya","Khamal"])
print(s)
```
