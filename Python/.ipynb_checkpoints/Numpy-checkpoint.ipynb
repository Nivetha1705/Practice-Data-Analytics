NUMPY - Numerical Python
Numpy is a combination of Numerics and Numarray created by Travis Oliphant in 2005
Numpy is built using C programming language so it is faster
C langueage is the fastest among all the programming languages

Python is slow when compared to Numpy. 
Numpy is fast.
The below example can be used to ensure that python is slow and numpy is faster as it was created using C programming language

```python
#Here it takes a lot of time to append each value in the lsit by adding 10 to it every time to the i value
l1=[10,20,30,40,50]
l2=[]
for i in l1:
    l2.append(i+10)
print(l2)
```

```python
#Instead we can use numpy where no loops or i values is needed
import numpy as np
a=np.array([10,20,30,40,50])
b=a+10
print(a)
print(b)
```
