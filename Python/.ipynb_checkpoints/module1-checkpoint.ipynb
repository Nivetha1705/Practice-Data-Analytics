```python
#We can access everything inside the module by importing it
import module1
print(module1.c)
module1.fun1()
module1.fun2()
```

```python
#We can access everything inside the classs by creating a object like below
import module1
c=module1.Calculator1()
print(c.a)
print(c.b)
c.add()
c.sub()
c1=module1.Calculator2()
c1.mul()
c1.div()
```

```python
#instead of giving module1, we can give the name as we need like m
import module1 as m
m.fun1()
m.fun2()
print(m.c)
print(m.d)
c=m.Calculator1()
print(c.a)
print(c.b)
c.add()
c.sub()
c1=m.Calculator2()
c1.mul()
c1.div()
```

```python
#We can import whatever we need only, and not the entire content from module1
from module1 import Calculator1
c=Calculator1()
print(c.a)
print(c.b)
c.add()
c.sub()
```

```python
from module1 import fun1,fun2,Calculator1
c=Calculator1()
print(c.a)
print(c.b)
c.add()
c.sub()
fun1()
fun2()
```
