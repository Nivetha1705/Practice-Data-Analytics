```python
# Without parameters, without returns
def add():
    print(10+20)
add()    
```

```python
# With parameters, without returns
def sub(a,b):
    return (a-b)
print(sub(40,10))    
```

```python
#Without parametes, with returns
def mul():
    return(89*34)
print(mul())
```

```python
# with parameters with ruturns
def mul(a,b):
    c=a*b
    return c
print(mul(5,3))
```

```python
#multiple functions in a single cell
# 1 - no input no output
def add():
    c=10+20
    print(c)
add()

# 2 - input , output
def add(a,b):
    return a+b
print(add(5,2))    

# 3 - input, no output
def mul(a,b):
    c=a*b
    print(c)
mul(6,6)    

# 4 - no input,output
def mul():
    return 5*6
result=mul()
print(result)
```

```python
def add():
    print(5+6)
def sub(a,b):
    print(a-b)
def mul():
    return 5*6
def div(a,b):
    return a/b
add()
sub(8,2)
result=mul()
print(result)  
result2=div(25,2)
print(result2)   
```

```python
def calculator():
    a=100+200
    b=300-25
    c=25*5
    d=100//10
    e=60/5
    f=4**2
    return a,b,c,d,e,f
result=calculator()
print(result)
```

```python
#LAMBDA FUNCTION

square=lambda x: x*x
print(square(5))
```

```python
# *args
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3, 4))   # 10
```

```python
#  **kwargs
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(name="Priya", age=25)
```

```python
#Combined example
def demo(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

demo(1, 2, name="Ravi", city="Pune")
```
