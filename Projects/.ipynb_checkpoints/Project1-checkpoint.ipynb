```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.mean()))
```

```python
df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

```python
#Snake Casing for all colum names
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.columns
```

```python
df = df.rename(columns = {"purchase_amount_(usd)" : "purchase_amount"})
df.columns
```

```python
#Create a column 'age_group'

labels=['Young Adult','Adult','Middle_Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age_group','age']].head(10)
```

```python
#Create column purchase_frequency_days - Convert textual frequency days into actual no of days as frequency purchases of the customer

frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df["purchase_frequency_days"] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```
